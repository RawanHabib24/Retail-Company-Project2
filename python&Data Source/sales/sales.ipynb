```python
import pandas as pd
from datetime import datetime
```

```python
sales_97 = pd.read_csv('Sales 1997.csv')
sales_98 = pd.read_csv('Sales 1998.csv')
```

```python
sales  = pd.concat([sales_97, sales_98], ignore_index=True)
```

```python
sales
```

```python
sales.isna().sum()
```

```python
sales.drop_duplicates(inplace=True)
```

```python
sales.dtypes
```

```python
sales['transaction_date'] = pd.to_datetime(sales['transaction_date'], format='%m/%d/%Y')
sales['stock_date'] = pd.to_datetime(sales['stock_date'], format='%m/%d/%Y')
```
