A data quality issue was also identified in the low_fat column, where a subset of products had a missing (NaN) value instead of a defined 0 or 1. Rather than treating these as equivalent to "not low fat," the missing values were treated as "unknown," since no reliable source (numeric flag or product name) confirmed the product's classification. Where the product name explicitly contained the phrase "Low Fat," the value was inferred and corrected to 1; the remaining unresolved records were flagged as "Unknown" for further review.

```python
products.loc[(products['low_fat'].isna() == True) & (products['product_name'].str.contains('Low Fat')),"low_fat"] = 1
```

```python
products.loc[(products['product_name'].str.contains('Low Fat')) & (products['low_fat'] == 0)]
```

```python
products.loc[(products['product_name'].str.contains('Low Fat')) & (products['low_fat'] == 0)].shape
```

A Data inconsistency was identified in 21 products where the product name contains "Low Fat" while the corresponding low_fat column value is recorded as 0 this represents a conflict between two fields that are expected to reflect the same attribute indicating a possible error in data entry or an update that was applied to one source but not the other.


```python
products.loc[(products['product_name'].str.contains('Low Fat')) & (products['low_fat'] == 0),"low_fat"] = 1
```

```python
products.isna().sum()
```

```python
products['low_fat'] = products['low_fat'].replace({0: 'no', 1: 'yes'})
```

```python
products.loc[products['low_fat'].isna() == True,'low_fat'] = 'Unknown'
```

```python
products.isna().sum()
```

```python
products.loc[products['recyclable'].isna() == True] = products.loc[products['recyclable'].isna() == True].fillna(0)
```

```python
products.drop_duplicates(inplace=True)
```

```python
Q1 = products['product_retail_price'].quantile(0.25)
Q3 = products['product_retail_price'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = products[(products['product_retail_price'] < lower_bound) | 
                     (products['product_retail_price'] > upper_bound)]
outliers
```
