```python
import pandas as pd
from datetime import datetime
```

```python
calendar = pd.read_csv('Calendar.csv')
calendar
```

```python
calendar.isna().sum()
```

```python
calendar.duplicated().sum()
```

```python
calendar.dtypes
```

```python
calendar['date'] = pd.to_datetime(calendar['date'],format='%m/%d/%Y')
calendar
```

```python
calendar.dtypes
```

```python
calendar['year'] = calendar['date'].dt.year
calendar['month'] = calendar['date'].dt.month
calendar['month_name'] = calendar['date'].dt.month_name()
calendar['day'] = calendar['date'].dt.day
calendar['day_of_week'] = calendar['date'].dt.day_of_week
calendar['week_of_year'] = calendar['date'].dt.isocalendar().week
calendar['quarter'] = calendar['date'].dt.quarter
```

```python
calendar
```

```python
calendar.dtypes
```

```python
calendar['quarter'] = calendar['quarter'].astype('str')
```

```python
calendar['quarter'] = 'Q-' + calendar['quarter']
```
