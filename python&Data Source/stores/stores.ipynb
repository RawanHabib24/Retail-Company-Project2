```python
import pandas as pd
from datetime import datetime
```

```python
stores = pd.read_csv('Stores.csv')
stores
```

```python
stores.isna().sum()
```

```python
stores.drop_duplicates(inplace=True)
```

```python
stores.dtypes
```

```python
stores['first_opened_date'] = pd.to_datetime(stores['first_opened_date'], format='%m/%d/%Y')
stores['last_remodel_date'] = pd.to_datetime(stores['last_remodel_date'], format='%m/%d/%Y')
```

```python
stores['store_city'].unique()
```

```python
stores['store_state'].unique()
```

```python
state_mapping = {
    'WA': 'Washington', 'CA': 'California', 'OR': 'Oregon',
    'BC': 'British Columbia', 'DF': 'Distrito Federal'
}
stores['store_state'] = stores['store_state'].replace(state_mapping)
```
